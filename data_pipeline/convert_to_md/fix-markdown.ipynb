### Xử lí lỗi:
- Xuống dòng (giữ structure từ PDF)
- Các dòng Công Báo (thường ở đầu trang)
- Thay thế kiểu bullet: x) bằng x.

```python
import os
import re

VIETNAMESE_LETTER = r"a-zA-ZÀ-ỹĐđ"

def join_broken_lines(text: str) -> str:
    lines = text.split('\n')
    joined_lines = []
    
    for line in lines:
        line = line.strip()
        if not line:
            if joined_lines and joined_lines[-1] != "":
                joined_lines.append("")
            continue
            
        if joined_lines and joined_lines[-1] != "":
            prev_line = joined_lines[-1].strip()
            
            # 1. Kiểm tra dấu ngắt câu của dòng trước
            # Lưu ý: Loại bỏ các dấu định dạng Markdown ở cuối dòng trước khi check punct
            clean_prev = prev_line.rstrip('*_ ')
            ends_with_punct = re.search(r'[.:!;?]$', clean_prev)
            
            # 2. Kiểm tra xem dòng hiện tại có phải khởi đầu mục mới không
            # Chặn: Điều 1, Chương I, Khoản 1, 1., a), - 
            is_new_section = re.match(r'^(Điều|Chương|Mục|Khoản|\d+\.|[a-zđĐ]\)|\-)', line, re.IGNORECASE)
            
            if not ends_with_punct and not is_new_section:
                # Xử lý nối text thông minh
                # Nếu cả 2 đều bọc trong **
                if prev_line.startswith("**") and prev_line.endswith("**") and line.startswith("**") and line.endswith("**"):
                    p_content = prev_line.strip("* ")
                    l_content = line.strip("* ")
                    joined_lines[-1] = f"**{p_content} {l_content}**"
                # Nếu cả 2 đều bọc trong _
                elif prev_line.startswith("_") and prev_line.endswith("_") and line.startswith("_") and line.endswith("_"):
                    p_content = prev_line.strip("_ ")
                    l_content = line.strip("_ ")
                    joined_lines[-1] = f"_{p_content} {l_content}_"
                else:
                    joined_lines[-1] = prev_line + " " + line
                continue
        
        joined_lines.append(line)
        
    return "\n".join(joined_lines)

```

```python
def main(in_path: str, out_path: str):
    # PHẢI ĐỌC FILE Ở ĐÂY
    if not os.path.exists(in_path):
        print(f"File not found: {in_path}")
        return

    with open(in_path, 'r', encoding='utf-8') as f:
        content = f.read()

    fixed_content = join_broken_lines(content)

    with open(out_path, 'w', encoding='utf-8') as f:
        f.write(fixed_content)

    print(f"Fixed: {os.path.basename(in_path)} -> {os.path.basename(out_path)}")
```
